# Environment check (2026 edition)
Run this notebook to verify that the main scientific Python packages are available.

```python
import sys
import numpy, scipy, matplotlib, pandas, sklearn, IPython

print('Python:', sys.version.split()[0])
for package in (numpy, scipy, matplotlib, pandas, sklearn, IPython):
    print(f'{package.__name__}: {package.__version__}')

print('\nEnvironment check completed. If imports succeeded, core packages are available.')
```
